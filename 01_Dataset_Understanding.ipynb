## What is a Dataset?


A dataset is a collection of related data organized in a structured or unstructured format that can be used to analyze information, identify patterns, make decisions, and build Machine Learning models

In simple words, a dataset is a collection of information about a particular subject

For example, a college collects information about students, such as their names, ages, study hours, and exam marks. All this collected information together forms a dataset

 Vijay	         21	   7	    92
 
 Bunny       	 19	   3	    67
 
 Raju	         20	   6	    78

This table is a dataset because it contains related information about multiple students

- Rows: Each row represents one student
- Columns: Each column represents an attribute, such as age or marks
- Values: Individual pieces of information, such as 20 or 85
- Features: Input information that can help explain or predict an outcome, such as study hours
- Target: The value we want to predict, such as exam marks


Real-World Examples:

Healthcare	  Age, symptoms, test results	             Analyze health outcomes


Banking	     Transactions, balances,customer details	Detect suspicious transactions

Why Is a Dataset Important in Data Analysis?

A dataset provides the information needed to answer questions and discover useful patterns

For example, by analyzing student marks, we can calculate average marks, identify the highest-scoring student, and investigate the relationship between study hours and performance

In AI and Machine Learning, datasets are used to train models, evaluate their performance, and make predictions on new data

Conclusion: A dataset is the foundation of data analysis. Understanding its structure, features, data types, and quality is the first step toward building reliable data analysis and Machine Learning solutions

## What is a Target?

```python
A target variable is the variable or outcome that we want to predict or explain using data analysis or a Machine Learning model.

In simple words, a target is the answer that we want our Machine Learning model to predict

For example, if we want to predict a student's exam marks using study hours, attendance, and age, then Marks is the target variable


Example:

Consider the following student dataset:
    
     Age	Study_Hours	Attendance	Marks
     20	         5	       90	    85
     21	         7	       95	    92
     19	         3	       75	    67
     20	         6	       85	    78

Suppose our goal is to predict student marks

- Input features: Age, Study_Hours, Attendance
- Target variable: Marks

The model learns patterns from the input features and their corresponding marks


Types of Target Variables:

A.Numerical target

A numerical target contains numeric values

Examples:

- House prices
- Student marks
- Employee salaries
- Monthly sales revenue

Use case: Predicting a house price of ₹50,00,000.

This is commonly called a regression problem

B.Categorical target

A categorical target represents a class or category

Examples:

- Customer churn: Yes or No
- Email: Spam or Not Spam
- Loan application: Approved or Rejected
- Disease classification: Positive or Negative

Use case: Predicting whether a customer will leave a company

This is commonly called a classification problem


Real-world business example:

Consider a telecommunications company that wants to identify customers who may cancel their subscriptions

    Customer_ID	Tenure	MonthlyCharges	Contract	 Churn

       C101	     2	      80.50	     Month-to-month	  Yes
       C102	     48	      65.20	       Two year	       No
       C103	      5	      90.10	     Month-to-month	  Yes
       C104	     36	      55.00	       One year	       No

Here:

- Tenure is a feature
- MonthlyCharges is a feature
- Contract is a feature
- Customer_ID is an identifier
- Churn is the target variable

The business uses historical customer information to train a model that predicts the likelihood of customer churn. 
The company can then investigate ways to retain customers identified as being at risk
```

```python
import pandas as pd

data = {
    "Age": [20, 21, 19, 20],
    "Study_Hours": [5, 7, 3, 6],
    "Attendance": [90, 95, 75, 85],
    "Marks": [85, 92, 67, 78]
}

df = pd.DataFrame(data)

X = df[["Age", "Study_Hours", "Attendance"]]

y = df["Marks"]

print("Input Features (X):")
print(X)

print("\nTarget Variable (y):")
print(y)
```

```python
Explanation:

- pd.DataFrame(data) creates a tabular dataset
- X stores the input features used for prediction
- y stores the target values the model should learn to predict
- df["Marks"] selects the Marks column as a Pandas Series

In a typical supervised learning workflow, X and y are then divided into training and testing sets
```

## Independent Variables


```python
An independent variable is a variable that is used to explain or predict changes in another variable

In data analysis and Machine Learning (ML), independent variables are commonly used as input features to predict a target variable

Simple definition: An independent variable is an input variable that may influence or help predict an outcome

For example, if we want to predict a student's exam marks, we might use study hours, attendance, and practice-test scores as independent variables

Example:

Imagine the following student dataset:
    
       StudyHours	Attendance(%)	PracticeTests	Exam Marks
    
           2	         60	               1	         45
           4	         75	               3	         65
           6	         85                5	         80
           8	         95	               8           	 92
    
In this example:
    
- Independent variables: Study Hours, Attendance, Practice Tests
- Dependent variable (target): Exam Marks
```

Independent Variables in Machine Learning:

In supervised Machine Learning, a dataset is commonly divided into two parts:

- X: Independent variables or input features
- y: Dependent variable or target to predict
  
The relationship can be represented as:
\[
y = f(X)
\]

Here:

- \(X\) represents the input features.
- \(f\) represents the model's learned relationship.
- \(y\) represents the target variable
 
For example, a model can use study hours, attendance, and practice tests to predict a student's exam marks.

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8],
    "Attendance": [60, 75, 85, 95],
    "Practice_Tests": [1, 3, 5, 8],
    "Exam_Marks": [45, 65, 80, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[["Study_Hours", "Attendance", "Practice_Tests"]]

print("Independent Variables:")
print(X)
```

```python
Explanation:

- df[...] selects columns from the DataFrame
- X stores the selected independent variables
- The double square brackets [[...]] select multiple columns and return a DataFrame
```

## Dependent Variables


```python
A dependent variable is the variable whose value we want to explain, understand, or predict using other variables

In Data Analysis and Machine Learning, the dependent variable is often called the target variable or output variable

Simple definition: A dependent variable is the outcome that may change depending on other variables

For example, if we want to predict a student's exam marks based on study hours and attendance, then Exam Marks is the dependent variable

Real-World Example:
    
Imagine that we have the following student dataset
    
     StudyHours	Attendance (%)	Practice Tests	Exam Marks
    
          2	         60	              1	           45
    
          4	         75	              3	           65
    
          6	         85	              5            80
    
          8	         95	              8	           92
    
In this dataset:
    
- Independent variables: Study Hours, Attendance, Practice Tests
- Dependent variable: Exam Marks
    
We use the independent variables to understand or predict the dependent variable

For example, we might investigate whether students who study longer tend to achieve higher marks
    This is an association to investigate, not proof that study hours alone cause higher marks


Types of Dependent Variables
    
There are two common types, based on the prediction task.

1.Numerical dependent variable

The target is a numerical value.

Examples:
- Exam marks: 85
- House price: 2500000
- Monthly sales: 50000
    
Usually associated with regression tasks

2.Categorical dependent variable

The target represents a category or class.

Examples:
- Customer churn: Yes or No
- Email: Spam or Not Spam
- Disease screening result: Positive or Negative
    
Usually associated with classification tasks
```

Dependent Variables in Machine Learning
    
In supervised Machine Learning, we usually divide the dataset into two parts:
    
- X — input features (independent variables).
- y — target values (dependent variable).
    
The relationship can be represented as:
\[
y = f(X)
\]

Where:
- \(X\) = input features
- \(f\) = relationship learned by the model
- \(y\) = actual target value

After training, a model uses new input features to predict the target. The predicted value is often written as \(\hat{y}\)

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8],
    "Attendance": [60, 75, 85, 95],
    "Practice_Tests": [1, 3, 5, 8],
    "Exam_Marks": [45, 65, 80, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
y = df["Exam_Marks"]                  

print("Dependent Variable:")
print(y)
```

```python
Explanation:

- df["Exam_Marks"] selects the Exam_Marks column
- y stores the target values
- print(y) displays the values that we want to predict
```

## Categorical Features

```python
Categorical features are dataset columns that represent groups, labels, classes, or categories rather than measurable quantities

They help us classify data into different groups and understand patterns among those groups

For example, in a customer dataset:
    
- Gender — Male, Female
- Contract — Monthly, Yearly
- PaymentMethod — Cash, Credit Card, Bank Transfer
- City — Hyderabad, Chennai, Bengaluru
    
These are categorical features because their values describe categories


Real-World Example:
    
Consider the following student dataset
    
      Student	Gender	Course	StudyMode	Exam Marks
    
         A	    Male	Python	 Online	      85
    
         B	   Female	Java	 Offline	  90
    
         C	   Male	    Python	 Offline	  75
    
         D	  Female	SQL	     Online	      88
    
         E	  Male	    Java	 Online	      80


In this dataset:
    
- Gender is a categorical feature.
- Course is a categorical feature.
- Study Mode is a categorical feature.
- Exam Marks is a numerical feature and could be the target if we want to predict marks.
    
We can use categorical features to answer questions such as:

- Which course do students prefer?
- How many students study online?
- Which study mode is associated with higher average marks?

These questions help us explore patterns, although associations alone do not prove causation

Types of Categorical Features

There are two common types of categorical features

A. Nominal features

Categories have no natural order.

Examples:

- Gender
- City
- Product Category
- Payment Method
- Internet Service

For example, Hyderabad, Chennai, and Bengaluru are city categories. One city is not inherently greater than another

B. Ordinal features

Categories have a meaningful order, but the distance between categories may not be equal.

Examples:

- Satisfaction: Low, Medium, High
- Education level: School, Bachelor's, Master's, Doctorate
- T-shirt size: Small, Medium, Large
- Service rating: Poor, Fair, Good, Excellent
For example, Low < Medium < High represents an order, but the difference between Low and Medium is not necessarily equal to the difference between Medium and High

Why Are Categorical Features Important?
    
1. Grouping data: Organize customers, products, students, or transactions into categories
2. Finding patterns: Compare counts, percentages, and outcomes across groups
3. Business insights: Discover which contract types or payment methods are associated with customer churn
4. Data visualization: Create bar charts and count plots
5. Machine Learning: Encode categories so compatible algorithms can use them
```

```python
import pandas as pd                  #Creating a sample dataset

data = {
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Course": ["Python", "Java", "Python", "SQL", "Java"],
    "Study_Mode": ["Online", "Offline", "Offline", "Online", "Online"],
    "Exam_Marks": [85, 90, 75, 88, 80]
}

df = pd.DataFrame(data)

print(df)
```

```python
categorical_features = df.select_dtypes(            #Identify categorical columns
    include=["object", "category", "string"]
)

print(categorical_features)
```

```python
Explanation:

- select_dtypes() selects columns based on their data types
- include=["object", "category", "string"] selects common Pandas text and categorical data types
- Exam_Marks is excluded because it is numeric
Note that numeric-coded categories, such as 1 = Low, 2 = Medium, and 3 = High, will not be selected by this code
    You must identify those based on their meaning
```

```python
print(df["Course"].value_counts())            #Count the categories
```

```python
Explanation:

- value_counts() counts how often each category appears
- Python appears twice, Java appears twice, and SQL appears once

This is useful for understanding customer preferences, product demand, or student enrollment
```

```python
print(df["Study_Mode"].value_counts(normalize=True) * 100)     #Calculate category percentages
```

## Target Variable

## Metadata

## Dataset Dimensions

```python
Dataset Dimensions refer to the number of rows and columns in a dataset

- Rows: Represent individual records, observations, or samples
- Columns: Represent features, variables, or attributes

In simple words, dataset dimensions tell us how many records and how many features a dataset contains

For a Pandas DataFrame, we can use the .shape attribute to find its dimensions


Real-World Example:
    
Imagine a dataset containing student information
    
        Student_ID	Age	Study_Hours	   Marks
    
           S001	     20	     2	        55
           S002	     21	     4	        70
           S003	     19	     6	        85
           S004	     22	     3	        62
           S005	     20	     5	        78
    
This dataset contains:
    
- 5 rows — five student records
- 4 columns — four variables
- Dataset dimensions: \(5 \times 4\)

Why Are Dataset Dimensions Important in Machine Learning?
    
Dataset dimensions help us understand the size and structure of our data before training a model
    
- Data validation: Check whether the dataset has the expected number of rows and columns
- Data cleaning: Verify whether records or columns were accidentally removed
- Feature selection: Understand how many potential input variables are available
- Model preparation: Check the size of the input data before training
- Data processing: Monitor changes after filtering, merging, or removing missing values
```

```python
import pandas as pd

data = {
    "Student_ID": ["S001", "S002", "S003", "S004", "S005"],
    "Age": [20, 21, 19, 22, 20],
    "Study_Hours": [2, 4, 6, 3, 5],
    "Marks": [55, 70, 85, 62, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.shape)  #Find the dataset dimensions
```

```python
Explanation:

- 5 represents the number of rows
- 4 represents the number of columns

Remember that .shape is an attribute, so you write df.shape, not df.shape()
```

```python
print(df.shape[0])       #Find the number of rows
```

```python
print(df.shape[1])     #Find the number of columns
```

```python
print(len(df))         # Number of rows
print(len(df.columns)) # Number of columns
```
