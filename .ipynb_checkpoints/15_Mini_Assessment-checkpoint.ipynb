#### Perform complete correlation analysis on the employee dataset in a single Python program.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("employee_eda_sample.csv")

# Select numerical features
numerical_columns = [
    "Age",
    "Experience_Years",
    "Salary",
    "Performance_Rating"
]

# Calculate Pearson correlation
correlation_matrix = df[numerical_columns].corr(method="pearson")

print("Correlation Matrix:")
display(correlation_matrix)

# Calculate Spearman correlation
spearman_matrix = df[numerical_columns].corr(method="spearman")

print("\nSpearman Correlation:")
display(spearman_matrix)

# Visualize correlation using heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Employee Feature Correlation Heatmap")
plt.show()

# Analyze Experience vs Salary
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Experience_Years",
    y="Salary"
)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()

# Calculate Experience vs Salary correlation
r = df["Experience_Years"].corr(df["Salary"])

print("\nExperience vs Salary Correlation:", round(r, 3))

# Interpret correlation
if r >= 0.7:
    print("Strong Positive Correlation")

elif r >= 0.3:
    print("Moderate Positive Correlation")

elif r > 0:
    print("Weak Positive Correlation")

elif r <= -0.7:
    print("Strong Negative Correlation")

elif r <= -0.3:
    print("Moderate Negative Correlation")

elif r < 0:
    print("Weak Negative Correlation")

else:
    print("No Linear Correlation")

# Find strongest positive and negative correlations
upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlation_pairs = upper_triangle.stack()

print("\nStrongest Positive Correlation:")
print(correlation_pairs.idxmax(), correlation_pairs.max())

print("\nStrongest Negative Correlation:")
print(correlation_pairs.idxmin(), correlation_pairs.min())
```

#### Analyze the relationship between employee experience and salary using the employee dataset.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("employee_eda_sample.csv")

# Select two variables
data = df[["Experience_Years", "Salary"]].dropna()

# Calculate correlation
correlation = data["Experience_Years"].corr(
    data["Salary"]
)

print("Correlation Coefficient:", round(correlation, 3))

# Visualize relationship
plt.figure(figsize=(8, 5))

sns.regplot(
    data=data,
    x="Experience_Years",
    y="Salary"
)

plt.title("Employee Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()

# Compare average salary across experience groups
data["Experience_Group"] = pd.cut(
    data["Experience_Years"],
    bins=[-np.inf, 2, 5, 10, np.inf],
    labels=["0-2 Years", "3-5 Years", "6-10 Years", "Above 10 Years"]
)

average_salary = data.groupby(
    "Experience_Group",
    observed=False
)["Salary"].mean()

print("\nAverage Salary by Experience Group:")
display(average_salary)

# Visualize average salary
average_salary.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Salary by Experience Group")
plt.xlabel("Experience Group")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.show()

# Final interpretation
if correlation > 0:
    print("\nExperience and Salary have a positive correlation.")
elif correlation < 0:
    print("\nExperience and Salary have a negative correlation.")
else:
    print("\nNo linear correlation was found.")

print(
    "\nImportant: Correlation does not prove that "
    "experience causes higher salary."
)
```

#### Identify highly correlated independent variables and calculate their VIF values.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# Load dataset
df = pd.read_csv("employee_eda_sample.csv")

# Select independent numerical features
features = [
    "Age",
    "Experience_Years",
    "Salary",
    "Performance_Rating"
]

# Handle missing and infinite values for this analysis
X = df[features].replace([np.inf, -np.inf], np.nan).dropna()

# Calculate correlation matrix
correlation_matrix = X.corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Visualize correlation heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Multicollinearity - Correlation Heatmap")
plt.show()

# Identify highly correlated feature pairs
print("\nHighly Correlated Feature Pairs:")

for i in range(len(features)):
    for j in range(i + 1, len(features)):

        correlation = correlation_matrix.iloc[i, j]

        if abs(correlation) >= 0.7:
            print(
                features[i],
                "and",
                features[j],
                "Correlation:",
                round(correlation, 3)
            )

# Calculate VIF
X_vif = add_constant(X)

vif_data = pd.DataFrame()

vif_data["Feature"] = X_vif.columns

vif_data["VIF"] = [
    variance_inflation_factor(X_vif.values, i)
    for i in range(X_vif.shape[1])
]

print("\nVariance Inflation Factor (VIF):")
display(vif_data)
```

#### Use the Attrition column as the target variable.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("employee_eda_sample.csv")

# Count observations in each class
class_counts = df["Attrition"].value_counts()

print("Class Counts:")
print(class_counts)

# Calculate class percentages
class_percentages = df["Attrition"].value_counts(normalize=True) * 100

print("\nClass Percentages:")
print(class_percentages.round(2))

# Display class distribution
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Attrition")

plt.title("Attrition Class Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()

# Check class imbalance
majority_class = class_counts.max()
minority_class = class_counts.min()

imbalance_ratio = majority_class / minority_class

print("\nImbalance Ratio:", round(imbalance_ratio, 2))

if imbalance_ratio >= 2:
    print("Class imbalance detected.")
else:
    print("No major class imbalance detected.")
```

#### Identify outliers in the employee dataset using the IQR method and examine whether they should be removed.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_eda_sample.csv")

features = ["Age", "Experience_Years", "Salary",
            "Performance_Rating"]

for col in features:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\nFeature: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))

    print(outliers[[col]])

    plt.figure(figsize=(7, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Outlier Analysis - {col}")
    plt.show()
```

#### Detect outliers in the employee dataset using the IQR method.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_eda_sample.csv")

numerical_features = [
    "Age",
    "Experience_Years",
    "Salary",
    "Performance_Rating"
]

for col in numerical_features:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print("\nFeature:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))

    print("Outlier Values:")
    print(outliers[["Employee_ID", col]])

    plt.figure(figsize=(7, 4))
    sns.boxplot(x=df[col])
    plt.title(f"IQR Outlier Detection - {col}")
    plt.show()
```

####  Analyze the distribution of Salary using descriptive statistics, histogram, KDE, and boxplot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_eda_sample.csv")

feature = "Salary"

salary = pd.to_numeric(df[feature], errors="coerce")
salary = salary.dropna()
salary = salary[salary.map(lambda x: pd.notna(x) and abs(x) != float("inf"))]

print("Distribution Analysis:", feature)

print("\nMean:", salary.mean())
print("Median:", salary.median())
print("Minimum:", salary.min())
print("Maximum:", salary.max())
print("Standard Deviation:", salary.std())
print("Variance:", salary.var())
print("Skewness:", salary.skew())
print("Kurtosis:", salary.kurt())

plt.figure(figsize=(8, 5))

sns.histplot(salary, kde=True, bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()

plt.figure(figsize=(8, 4))

sns.boxplot(x=salary)

plt.title("Salary Distribution - Boxplot")
plt.xlabel("Salary")

plt.show()
```

#### Generate a complete EDA summary.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD DATASET
df = pd.read_csv("employee_eda_sample.csv")

# Convert Joining_Date into datetime
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"], errors="coerce"
)

print("=" * 50)
print("1. DATASET OVERVIEW")
print("=" * 50)

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst Five Records:")
print(df.head())

print("\nDataset Information:")
df.info()

# 2. DATA QUALITY ANALYSIS
print("\n" + "=" * 50)
print("2. DATA QUALITY ANALYSIS")
print("=" * 50)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print((df.isnull().mean() * 100).round(2))

print("\nDuplicate Records:", df.duplicated().sum())

print("\nDuplicate Employee IDs:",
      df["Employee_ID"].duplicated().sum())

print("\nNegative Values:")
for col in ["Age", "Experience_Years", "Salary"]:
    print(col, ":", (df[col] < 0).sum())

print("\nInvalid Joining Dates:",
      df["Joining_Date"].isnull().sum())

print("\nUnexpected Departments:")
print(df.loc[
    ~df["Department"].isin(
        ["IT", "HR", "Finance", "Sales"]
    ) & df["Department"].notna(),
    "Department"
].unique())

# 3. NUMERICAL ANALYSIS
print("\n" + "=" * 50)
print("3. NUMERICAL ANALYSIS")
print("=" * 50)

numerical_features = [
    "Age",
    "Experience_Years",
    "Salary",
    "Performance_Rating"
]

print(df[numerical_features].describe())

print("\nSkewness:")
print(df[numerical_features].skew())

print("\nKurtosis:")
print(df[numerical_features].kurt())

# 4. OUTLIER DETECTION USING IQR
print("\n" + "=" * 50)
print("4. OUTLIER DETECTION")
print("=" * 50)

for col in numerical_features:

    data = df[col].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower) | (data > upper)
    ]

    print(f"\nFeature: {col}")
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
    print("Outlier Count:", len(outliers))
    print("Outlier Values:", outliers.tolist())

# 5. CATEGORICAL ANALYSIS
print("\n" + "=" * 50)
print("5. CATEGORICAL ANALYSIS")
print("=" * 50)

categorical_features = [
    "Department",
    "Gender",
    "Attrition"
]

for col in categorical_features:

    print(f"\nFeature: {col}")

    print("\nCategory Frequency:")
    print(df[col].value_counts(dropna=False))

    print("\nCategory Percentage:")
    print(
        (df[col].value_counts(normalize=True,
                              dropna=False) * 100).round(2)
    )

# 6. TARGET VARIABLE ANALYSIS
print("\n" + "=" * 50)
print("6. TARGET VARIABLE ANALYSIS")
print("=" * 50)

print("\nAttrition Distribution:")
print(df["Attrition"].value_counts())

print("\nAttrition Percentage:")
print(
    (df["Attrition"].value_counts(normalize=True) * 100)
    .round(2)
)

# 7. CORRELATION ANALYSIS
print("\n" + "=" * 50)
print("7. CORRELATION ANALYSIS")
print("=" * 50)

correlation = df[numerical_features].corr()

print(correlation.round(2))

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Employee Dataset - Correlation Heatmap")
plt.tight_layout()
plt.show()

# 8. DISTRIBUTION VISUALIZATION
print("\n" + "=" * 50)
print("8. DISTRIBUTION ANALYSIS")
print("=" * 50)

for col in numerical_features:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        data=df,
        x=col,
        kde=True,
        bins=10
    )

    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()

# 9. BOXPLOT ANALYSIS
for col in numerical_features:

    plt.figure(figsize=(7, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")
    plt.tight_layout()
    plt.show()

# 10. CATEGORICAL VISUALIZATION
for col in categorical_features:

    plt.figure(figsize=(7, 4))

    sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index
    )

    plt.title(f"Category Distribution - {col}")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# 11. DEPARTMENT VS ATTRITION
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nDepartment-wise Attrition (%):")
print(department_attrition.round(2))

department_attrition.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Department-wise Attrition")
plt.ylabel("Percentage")
plt.tight_layout()
plt.show()

# 12. FINAL EDA SUMMARY
print("\n" + "=" * 50)
print("12. FINAL EDA SUMMARY")
print("=" * 50)

print("Total Records:", df.shape[0])
print("Total Features:", df.shape[1])
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())
print("Target Variable: Attrition")

print("\nEDA Completed Successfully!")
```
