# Notebook 4 – Univariate Analysis

Univariate analysis is the process of analyzing a single variable to understand its distribution, central value, and range of values

For example, if a dataset contains Age, Salary, Experience, and Department, we analyze each numerical column Separately

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [22, 25, 28, 25, 30, 35, 40, 25, 32, 28],
    "Salary": [20000, 25000, 30000, 25000, 35000,
               40000, 50000, 25000, 32000, 45000],
    "Experience": [1, 2, 3, 2, 5, 8, 10, 2, 6, 7]
})

df.head()
```

```python
Interpret the results:
    
- Mean: The average age is 29 years
- Median: Half of the observations are at or below 28 years
- Mode: The most frequently occurring age is 25
- Minimum and maximum: Ages range from 22 to 40 years
- Range: The difference between the highest and lowest ages is 18 years
- Variance: The sample variance is approximately 32.89 years²
- Standard deviation: Ages have a sample standard deviation of approximately 5.74 years
- Quartiles: The middle 50% of observations lie between Q1 = 25 and Q3 = 32
- Percentiles: The 90th percentile is 35.5, meaning approximately 90% of observations fall at or below that value under Pandas' interpolation method
- Skewness: The positive value indicates some right-skewness in this sample
- Kurtosis: The excess kurtosis is close to zero, but this small sample is not sufficient to conclude that the distribution is normal
```

### Histogram

A histogram shows the frequency distribution of a numerical variable by grouping values into intervals called bins.

```python
import matplotlib.pyplot as plt

age = df["Age"].dropna()

plt.figure(figsize=(8, 5))
plt.hist(age, bins=5, edgecolor="black")

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### KDE / Density Plot

A Kernel Density Estimation (KDE) plot provides a smooth estimate of the distribution of a numerical variable

```python
import seaborn as sns
import matplotlib.pyplot as plt

age = df["Age"].dropna()

plt.figure(figsize=(8, 5))
sns.kdeplot(data=age, fill=True)

plt.title("KDE / Density Plot of Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()
```

### Box Plot

A box plot summarizes the distribution using the median, quartiles, whiskers, and potential outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

age = df["Age"].dropna()

plt.figure(figsize=(8, 4))
sns.boxplot(x=age)

plt.title("Box Plot of Age")
plt.xlabel("Age")
plt.show()
```

### Categorical Variables


```python
import pandas as pd

df = pd.DataFrame({
    "Department": [
        "IT", "HR", "IT", "Sales", "HR",
        "IT", "Sales", "IT", "HR", "Sales"
    ]
})

df
```

### Frequency
Frequency analysis counts how many observations belong to each category.

```python
department = df["Department"]

print("Total observations:", department.count())
print("Frequency of each category:")
print(department.value_counts())
```

### Unique values

Unique values are the distinct categories present in a variable

```python
print("Unique values:", department.unique())
print("Number of unique values:", department.nunique())
```

### Value counts

value_counts() provides the count of observations for each category

```python
department.value_counts()
```

### Percentage distribution

Percentage distribution shows the proportion of observations belonging to each category.

Formula:

\[
\text{Percentage}=\frac{\text{Category Frequency}}{\text{Total Observations}}\times100
\]

```python
percentage = department.value_counts(normalize=True) * 100

print(percentage.round(2))
```

## Visualizations

### Bar Chart

A bar chart displays the frequency of each category using separate bars

```python
import matplotlib.pyplot as plt

counts = df["Department"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(counts.index, counts.values, edgecolor="black")

plt.title("Department Frequency")
plt.xlabel("Department")
plt.ylabel("Frequency")
plt.show()
```

```python
What to observe:
    
- Which category has the tallest bar?
- Which categories have similar frequencies?
- Are some categories rare or underrepresented?
    
Findings: IT has the tallest bar, with 4 observations. HR and Sales have equal frequencies of 3
```

### Count Plot

A count plot uses individual observations to calculate and display the frequency of each category. Seaborn's countplot() is convenient for this

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Department",
    order=df["Department"].value_counts().index
)

plt.title("Count Plot of Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
What to observe:
    
- Number of records in each category
- The most frequent category
- Differences in counts across categories

Findings: The count plot confirms that IT occurs 4 times, while HR and Sales occur 3 times each.

Bar chart vs. count plot: 
    
A bar chart usually displays values you have already calculated.
A count plot counts individual observations automatically. For this simple categorical example, both show similar information.
```

### Pie Chart

A pie chart displays categories as parts of a whole. Use it when the variable has a small number of categories and the proportions are easy to compare.

```python
counts = df["Department"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    counts.values,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Department Percentage Distribution")
plt.show()
```

```python
What to observe:
    
- Which category occupies the largest share?
- What proportion belongs to each category?
- Are the proportions balanced or dominated by one category?
    
Findings: IT accounts for 40% of the records. HR and Sales each account for 30%. No single department dominates the sample
```
