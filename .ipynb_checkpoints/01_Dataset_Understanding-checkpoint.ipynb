# Notebook 1 – Dataset Understanding


## What is a Dataset?


A dataset is a collection of related data organized in a structured or unstructured format that can be used to analyze information, identify patterns, make decisions, and build Machine Learning models

In simple words, a dataset is a collection of information about a particular subject

For example, a college collects information about students, such as their names, ages, study hours, and exam marks. All this collected information together forms a dataset

 StudentName	Age	StudyHours	Marks

 Ajay	         20	   5	    85
 
 Vijay	         21	   7	    92
 
 Bunny       	 19	   3	    67
 
 Raju	         20	   6	    78
 
 Arun	         22	   2	    55


This table is a dataset because it contains related information about multiple students

- Rows: Each row represents one student
- Columns: Each column represents an attribute, such as age or marks
- Values: Individual pieces of information, such as 20 or 85
- Features: Input information that can help explain or predict an outcome, such as study hours
- Target: The value we want to predict, such as exam marks


Real-World Examples:

Domain	       Datasetinformation	                      Purpose


Healthcare	  Age, symptoms, test results	             Analyze health outcomes


Banking	     Transactions, balances,customer details	Detect suspicious transactions


E-commerce	 Products, prices, purchases	            Understand customer behavior


Why Is a Dataset Important in Data Analysis?

A dataset provides the information needed to answer questions and discover useful patterns

For example, by analyzing student marks, we can calculate average marks, identify the highest-scoring student, and investigate the relationship between study hours and performance

In AI and Machine Learning, datasets are used to train models, evaluate their performance, and make predictions on new data

Conclusion: A dataset is the foundation of data analysis. Understanding its structure, features, data types, and quality is the first step toward building reliable data analysis and Machine Learning solutions

```python
import pandas as pd

data = {
    "Name": ["Ajay", "Vijay", "Bunny", "Bunny"],
    "Age": [20, 21, 19, 20],
    "Marks": [85, 92, 67, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
Explanation:

- import pandas as pd imports the Pandas library
- data stores the student information in a dictionary
- pd.DataFrame(data) converts the dictionary into a tabular DataFrame
- print(df) displays the dataset
```

## What is a Feature?


A feature is an individual attribute, property, or characteristic of an observation in a dataset. Features provide information that helps us understand data, identify patterns, and make predictions using Machine Learning

In simple words, a feature is a column in a dataset that describes something about the data

For example, in a student dataset, Age, Study_Hours, and Attendance are features because they describe different characteristics of each student

Example:

Consider the following student dataset:

     Name	Age	Study_Hours	Attendance	Marks
     
     Ajay	20	   5	     90	        85
     
     Vijay	21	   7	     95	        92
     
     Bunny	19	   3	     75	        67
     
     Raju	20	   6	     85	        78


In this dataset:

- Age is a feature
- Study_Hours is a feature
- Attendance is a feature
- Marks can be the target variable if we want to predict student marks
- Name identifies each student and usually isn't a useful predictive feature
- 
Features can be numerical or categorical, depending on the type of information they represent


Types of Features:

A.Numerical features

Numerical features contain numeric values representing quantities or measurements

Examples:

- Age
- Salary
- Study hours
- Monthly charges
- 
B. Categorical features

Categorical features represent groups or categories

Examples:

- Gender
- City
- Contract type
- Payment method
- 
Categorical features often need to be encoded before they can be used by machine learning algorithms


Real-world example:

Consider an e-commerce company that wants to understand customer purchases

     Customer_Age	Monthly_Income	Previous_Orders	Membership
     
          25	       30000	           4	     Gold
          35	       50000	          10	    Silver
          28	       35000	           6	    Gold
          42	       60000	          15	    Platinum
          
Here, Customer_Age, Monthly_Income, Previous_Orders, and Membership are features

The company can analyze these features to understand purchasing patterns and customer behavior

```python
import pandas as pd

data = {
    "Age": [20, 21, 19, 20],
    "Study_Hours": [5, 7, 3, 6],
    "Attendance": [90, 95, 75, 85],
    "Marks": [85, 92, 67, 78]
}

df = pd.DataFrame(data)

print(df)

X = df[["Age", "Study_Hours", "Attendance"]]

print("\nFeatures:")
print(X)

y = df["Marks"]

print("\nTarget:")
print(y)
```

```python
Explanation:

- pd.DataFrame(data) creates a dataset
- df[["Age", "Study_Hours", "Attendance"]] selects multiple feature columns
- df["Marks"] selects the target column
- X contains the input features
- y contains the target values
```

## What is a Target?

```python
A target variable is the variable or outcome that we want to predict or explain using data analysis or a Machine Learning model.

In simple words, a target is the answer that we want our Machine Learning model to predict

For example, if we want to predict a student's exam marks using study hours, attendance, and age, then Marks is the target variable


Example:

Consider the following student dataset:
    
     Age	Study_Hours	Attendance	Marks
     20	         5	       90	    85
     21	         7	       95	    92
     19	         3	       75	    67
     20	         6	       85	    78

Suppose our goal is to predict student marks

- Input features: Age, Study_Hours, Attendance
- Target variable: Marks

The model learns patterns from the input features and their corresponding marks


Types of Target Variables:

A.Numerical target

A numerical target contains numeric values

Examples:

- House prices
- Student marks
- Employee salaries
- Monthly sales revenue

Use case: Predicting a house price of ₹50,00,000.

This is commonly called a regression problem

B.Categorical target

A categorical target represents a class or category

Examples:

- Customer churn: Yes or No
- Email: Spam or Not Spam
- Loan application: Approved or Rejected
- Disease classification: Positive or Negative

Use case: Predicting whether a customer will leave a company

This is commonly called a classification problem


Real-world business example:

Consider a telecommunications company that wants to identify customers who may cancel their subscriptions

    Customer_ID	Tenure	MonthlyCharges	Contract	 Churn

       C101	     2	      80.50	     Month-to-month	  Yes
       C102	     48	      65.20	       Two year	       No
       C103	      5	      90.10	     Month-to-month	  Yes
       C104	     36	      55.00	       One year	       No

Here:

- Tenure is a feature
- MonthlyCharges is a feature
- Contract is a feature
- Customer_ID is an identifier
- Churn is the target variable

The business uses historical customer information to train a model that predicts the likelihood of customer churn. 
The company can then investigate ways to retain customers identified as being at risk
```

```python
import pandas as pd

data = {
    "Age": [20, 21, 19, 20],
    "Study_Hours": [5, 7, 3, 6],
    "Attendance": [90, 95, 75, 85],
    "Marks": [85, 92, 67, 78]
}

df = pd.DataFrame(data)

X = df[["Age", "Study_Hours", "Attendance"]]

y = df["Marks"]

print("Input Features (X):")
print(X)

print("\nTarget Variable (y):")
print(y)
```

```python
Explanation:

- pd.DataFrame(data) creates a tabular dataset
- X stores the input features used for prediction
- y stores the target values the model should learn to predict
- df["Marks"] selects the Marks column as a Pandas Series

In a typical supervised learning workflow, X and y are then divided into training and testing sets
```

## Independent Variables


```python
An independent variable is a variable that is used to explain or predict changes in another variable

In data analysis and Machine Learning (ML), independent variables are commonly used as input features to predict a target variable

Simple definition: An independent variable is an input variable that may influence or help predict an outcome

For example, if we want to predict a student's exam marks, we might use study hours, attendance, and practice-test scores as independent variables

Example:

Imagine the following student dataset:
    
       StudyHours	Attendance(%)	PracticeTests	Exam Marks
    
           2	         60	               1	         45
           4	         75	               3	         65
           6	         85                5	         80
           8	         95	               8           	 92
    
In this example:
    
- Independent variables: Study Hours, Attendance, Practice Tests
- Dependent variable (target): Exam Marks
```

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8],
    "Attendance": [60, 75, 85, 95],
    "Practice_Tests": [1, 3, 5, 8],
    "Exam_Marks": [45, 65, 80, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[["Study_Hours", "Attendance", "Practice_Tests"]]

print("Independent Variables:")
print(X)
```

```python
Explanation:

- df[...] selects columns from the DataFrame
- X stores the selected independent variables
- The double square brackets [[...]] select multiple columns and return a DataFrame
```
