### Number of Rows


```python
The number of rows in a dataset represents how many records or observations the dataset contains

In Pandas, each row usually represents one observation, such as one student, one customer, one product, or one transaction

For example, in a student dataset:
    
           Student_ID	Name	Age 	Exam_Score

               101	    Ajay	20	        55
               102	    Vijay	21	        88
               103	    Bunny	19	        62
               104	    Raju	22	        95
               105	    Kiran	20	        72

This dataset contains 5 rows, representing 5 students
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Ajay", "Vijay", "Bunny", "Raju", "Kiran"],
    "Age": [20, 21, 19, 22, 20],
    "Exam_Score": [55, 88, 62, 95, 72]
}

df = pd.DataFrame(data)

df
```

```python
number_of_rows = len(df)            #Find the number of rows Using len()

print("Number of rows:", number_of_rows)
```

```python
number_of_rows = df.shape[0]        #Using shape()

print("Number of rows:", number_of_rows)
```

```python
number_of_rows = df.index.size       #Using Index()

print("Number of rows:", number_of_rows)
```

### Data Types

```python
Data types describe the kind of data stored in each column of a dataset.
    In Pandas, understanding data types helps you choose the correct operations, identify data quality problems, and prepare data for machine learning.

For example, a student's age is numerical, a student's name is text, and an exam date is a date/time value

          Student_ID	Name	Age	Study_Hours	Exam_Date	Result
          
                101	    Ravi	20	      2.5	2026-10-01	Fail
                102	    Priya	21        5.0	2026-10-02	Pass
                103	    Arun	19	      3.0	2026-10-03	Fail
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Study_Hours": [2.5, 5.0, 3.0, 6.5],
    "Passed": [False, True, False, True],
    "Exam_Date": [
        "2026-10-01",
        "2026-10-02",
        "2026-10-03",
        "2026-10-04"
    ]
}

df = pd.DataFrame(data)

df["Exam_Date"] = pd.to_datetime(df["Exam_Date"])

df
```

```python
print(df.dtypes)
```

```python
print(df["Age"].dtype)
print(df["Study_Hours"].dtype)
print(df["Name"].dtype)
print(df["Exam_Date"].dtype)
```

```python
df["Age"] = df["Age"].astype("int64")

print(df["Age"].dtype)
```

### Index


```python
An index in Pandas is used to label and identify the rows of a DataFrame or Series

By default, Pandas assigns an integer index starting from 0. However, you can also use meaningful labels, such as student IDs, dates, or product codes, as an index.

Example:
    
             Index	Student_ID	Name	Age	Exam_Score

                0	101	        Ravi	20	  55

                1	102     	Priya	21	  88

                2	103        	Arun	19	  62

                3	104	        Sneha	22	  95

Here, 0, 1, 2, and 3 are the row index labels

Important: The index is not necessarily a column. It is a separate part of the DataFrame structure
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Exam_Score": [55, 88, 62, 95]
}

df = pd.DataFrame(data)

df
```

```python
print(df.index)
```

```python
Explanation:

- start=0: The index begins at 0
- stop=4: The index ends before 4
- step=1: The index increases by 1 each time
```

### head()

```python
The head() method in Pandas is used to display the first few rows of a DataFrame or Series.

It is commonly used during data inspection to understand the structure of a dataset, check column names, and 
preview the values after loading a CSV or Excel file

Syntax:

df.head(n)

df — the DataFrame name
head() — displays the first rows
n — the number of rows to display. It is optional
Default value of n is 5
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Divya", "Rahul"],
    "Age": [20, 21, 19, 22, 20, 21, 23],
    "Study_Hours": [2.5, 5.0, 3.0, 6.0, 4.0, 5.5, 2.0],
    "Exam_Score": [55, 88, 62, 95, 72, 91, 48]
}

df = pd.DataFrame(data)

df
```

```python
df.head()
```

```python
Explanation: Since we did not specify a number inside head(), Pandas displays the first five rows by default
```

```python
df.head(3)
```

### tail()


```python
The tail() method in Pandas is used to display the last few rows of a DataFrame or Series

It is useful during data inspection to check the final records, verify that a file loaded correctly, and 
identify unusual values near the end of a dataset

Syntax:

df.tail(n)

df — the DataFrame name
tail() — displays the last rows
n — the number of rows to display; it is optional
The default value of n is 5
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Divya", "Rahul"],
    "Age": [20, 21, 19, 22, 20, 21, 23],
    "Study_Hours": [2.5, 5.0, 3.0, 6.0, 4.0, 5.5, 2.0],
    "Exam_Score": [55, 88, 62, 95, 72, 91, 48]
}

df = pd.DataFrame(data)

df
```

```python
df.tail()    #To Display last 5 rows
```

```python
df.tail(3)         #To Display Specific no of rows
```

### sample()

```python
In Pandas, the sample() method is used to select random rows from a DataFrame
It helps us inspect different parts of a dataset without displaying every row

For example, if a dataset contains 1,000 student records, sample() can display 5 randomly selected students.

Why is sample() useful in Data Inspection?

- Inspect random records from a dataset.
- Identify unusual or unexpected values.
- Check whether data appears consistent across different records.
- Quickly examine large datasets.

Syntax:

DataFrame.sample(n=5, random_state=None)


Parameter	        Description

n	              Number of random rows to select. Default is 1

random_state	Controls randomness so the same sample can be reproduced
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Ravi", "Priya", "Arun", "Sneha",
             "Kiran", "Divya", "Rahul", "Anu"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19],
    "Study_Hours": [3, 5, 2, 6, 4, 7, 3, 5],
    "Exam_Score": [65, 85, 55, 90, 72, 95, 68, 88]
}

df = pd.DataFrame(data)

df
```

```python
df.sample()     #To select one row
```

```python
df.sample(n=3)       #To select 3 rows
```

```python
df.sample(n=3, random_state=42)        #Random Sample
```

```python
print("Dataset shape:", df.shape)

print("\nRandom 5 rows:")
display(df.sample(n=5, random_state=42))

print("\nRandom 3 rows:")
display(df.sample(n=3, random_state=42))
```

### shape

```python
In Pandas, the shape attribute returns the number of rows and columns in a DataFrame

It helps us understand the dimensions of a dataset immediately after loading it

For example, if a dataset contains 100 student records and 5 columns, its shape is:
    

Shape = (100, 5)

- 100 represents the number of rows
- 5 represents the number of columns

Important: shape is an attribute, not a method. Therefore, use df.shape, not df.shape()

Syntax:

DataFrame.shape
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran"],
    "Age": [20, 21, 19, 22, 20],
    "Study_Hours": [3, 5, 2, 6, 4],
    "Exam_Score": [65, 85, 55, 90, 72]
}

df = pd.DataFrame(data)

df
```

```python
df.shape     #Find the dataset shape
```

```python
Explanation:

- 5 rows represent five student records
- 5 columns represent Student_ID, Name, Age, Study_Hours, and Exam_Score
```

### info()

```python
In Pandas, the info() method provides a summary of a DataFrame's structure and information

It helps us understand:
    
- Total number of rows and columns
- Column names
- Number of non-null values in each column
- Data types of columns
- Memory usage of the DataFrame
Real-world example: When you load a student dataset containing thousands of records,
info() helps you quickly check whether columns have missing values and whether their data types are appropriate.

Syntax:

DataFrame.info()
```

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran"],
    "Age": [20, 21, None, 22, 20],
    "Study_Hours": [3.5, 5.0, 2.0, None, 4.0],
    "Exam_Score": [65, 85, 55, 90, 72]
}

df = pd.DataFrame(data)

df
```

```python
df.info()
```

```python
print("Dataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
```
